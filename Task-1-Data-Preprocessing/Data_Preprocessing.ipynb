```python
# Codveda Machine Learning Internship
# Level 1 - Task 1
# Data Preprocessing for Machine Learning

import pandas as pd
import numpy as np

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler, LabelEncoder

# 1. Load the dataset

df = pd.read_csv(r"C:\Users\250 G7\Desktop\CodeVeda Intern\Dataset\1) iris.csv")

print("Dataset Shape:", df.shape)
print("\nFirst 5 Rows:")
print(df.head())

# 2. Check missing values

print("\nMissing Values:")
print(df.isnull().sum())

# Handle missing values
df = df.dropna()

print("\nShape after handling missing values:", df.shape)

# 3. Encode categorical target variable

le = LabelEncoder()

df["species_encoded"] = le.fit_transform(df["species"])

print("\nEncoded Classes:")
for original, encoded in zip(le.classes_, range(len(le.classes_))):
    print(original, "->", encoded)

# 4. Separate features and target

X = df.drop(columns=["species", "species_encoded"])
y = df["species_encoded"]

print("\nFeatures:")
print(X.columns.tolist())

print("\nTarget:")
print("species_encoded")

# 5. Standardize numerical features

scaler = StandardScaler()

X_scaled = scaler.fit_transform(X)

# Convert back to DataFrame for easier viewing
X_scaled = pd.DataFrame(
    X_scaled,
    columns=X.columns
)

print("\nStandardized Features:")
print(X_scaled.head())

# 6. Split dataset into training and testing

X_train, X_test, y_train, y_test = train_test_split(
    X_scaled,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

# 7. Display final dataset information


print("PREPROCESSING COMPLETED")


print("Training Features:", X_train.shape)
print("Testing Features :", X_test.shape)
print("Training Target  :", y_train.shape)
print("Testing Target   :", y_test.shape)

# 8. Visualizations

import matplotlib.pyplot as plt
import seaborn as sns

# A. Missing Values

plt.figure(figsize=(8, 5))

sns.heatmap(
    df.isnull(),
    cbar=False,
    yticklabels=False,
    cmap="viridis"
)

plt.title("Missing Values Heatmap")
plt.tight_layout()
plt.savefig("missing_values_heatmap.png", dpi=300)
plt.show()


# B. Feature Distributions Before Scaling

features = [
    "sepal_length",
    "sepal_width",
    "petal_length",
    "petal_width"
]

df[features].hist(
    figsize=(10, 7),
    bins=15
)

plt.suptitle("Feature Distributions Before Standardization")
plt.tight_layout()
plt.savefig("feature_distributions_before.png", dpi=300)
plt.show()


# C. Feature Distributions After Scaling

X_scaled[features].hist(
    figsize=(10, 7),
    bins=15
)

plt.suptitle("Feature Distributions After Standardization")
plt.tight_layout()
plt.savefig("feature_distributions_after.png", dpi=300)
plt.show()


# D. Class Distribution

plt.figure(figsize=(7, 5))

sns.countplot(
    x=df["species"]
)

plt.title("Class Distribution")
plt.xlabel("Species")
plt.ylabel("Number of Samples")
plt.tight_layout()
plt.savefig("class_distribution.png", dpi=300)
plt.show()
```
